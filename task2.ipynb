```python
import sqlite3
import pandas as pd

# 1. Load the uploaded CSV file into Python
df = pd.read_csv('SQL_Sales_Dataset_200_Rows.xlsx - Sheet1.csv')

# 2. Create an in-memory SQL database
conn = sqlite3.connect(':memory:')
df.to_sql('sales', conn, index=False, if_exists='replace')

# 3. Define the SQL Query as a Python string
query = """
SELECT
    customer_name,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(quantity) AS total_items,
    ROUND(SUM(total_price), 2) AS total_spent,
    ROUND(AVG(total_price), 2) AS avg_item_spend
FROM sales
GROUP BY customer_name
ORDER BY total_spent DESC
LIMIT 10;
"""

# 4. Execute the SQL query and display results
top_customers = pd.read_sql_query(query, conn)
print(top_customers)
```

```python
# Query: Category Summary
q_category = """
SELECT
    category,
    COUNT(order_id) AS total_orders,
    SUM(quantity) AS total_quantity_sold,
    ROUND(SUM(total_price), 2) AS total_revenue,
    ROUND(AVG(unit_price), 2) AS avg_unit_price
FROM sales
GROUP BY category
ORDER BY total_revenue DESC;
"""
print(pd.read_sql_query(q_category, conn))
```

```python
# Query: High vs Standard Price Orders
q_price_bracket = """
SELECT
    CASE
        WHEN unit_price >= 2000 THEN 'High Tier (>= $2,000)'
        WHEN unit_price >= 1000 THEN 'Mid Tier ($1,000 - $1,999)'
        ELSE 'Budget Tier (< $1,000)'
    END AS price_tier,
    COUNT(order_id) AS order_count,
    ROUND(SUM(total_price), 2) AS total_sales
FROM sales
GROUP BY price_tier
ORDER BY total_sales DESC;
"""
print(pd.read_sql_query(q_price_bracket, conn))
```
